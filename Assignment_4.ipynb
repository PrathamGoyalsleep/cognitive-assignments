```python
x=np.array([1,2,3,4,5,1,2,1,1,1])
values,count=np.unique(x,return_counts=True)

most_frequent_x=values[np.argmax(count)]
indices_x=np.where(x==most_frequent_x)

print("x=",x)
print("The most frequent value in the array is:",most_frequent_x)
print("indices:",indices_x[0])



y=np.array([1,1,1,2,3,4,2,4,3,3,3])
values,count=np.unique(y,return_counts=True)

most_frequent_y=values[np.argmax(count)]
indices_y=np.where(y==most_frequent_y)

print("y=",y)
print("The most frequent value in the array is:",most_frequent_y)
print("indices:",indices_y[0])
```

Q4

```python
import numpy as np

Pratham = np.linspace(10, 100, 25)

print("Array:")
print(Pratham)

print("Dimensions (ndim):", Pratham.ndim)

print("Shape:", Pratham.shape)

print("Total elements (size):", Pratham.size)

print("Data Type:", Pratham.dtype)

print("Total bytes consumed (nbytes):", Pratham.nbytes)

print("Transpose of 1-D array:")
print(Pratham.T)

column_array = Pratham.reshape(25, 1)

print("Reshaped as column:")
print(column_array)

print("Transpose of column_array:")
print(column_array.T)
```

```python
import numpy as np

ucs420_pratham = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original 2-D array:")
print(ucs420_pratham)

print("Mean:", np.mean(ucs420_pratham))

print("Median:", np.median(ucs420_pratham))

print("Maximum:", np.max(ucs420_pratham))

print("Minimum:", np.min(ucs420_pratham))

print("Unique elements:", np.unique(ucs420_pratham))

reshaped_ucs420_pratham = ucs420_pratham.reshape(4, 3)

print("Reshaped array (4*3):")
print(reshaped_ucs420_pratham)

resized_ucs420_pratham = np.resize(ucs420_pratham, (2, 3))

print("Resized array (2*3):")
print(resized_ucs420_pratham)
```
